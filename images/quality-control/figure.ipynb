```python
import numpy as np
from matplotlib import pyplot as plt

PRIME_BLUE = "#004d9b"
PRIME_LGREY = "#cbcfd2"
PRIME_DGREY = "#7e898f"
SECOND_DBLUE = "#002552"
SECOND_LBLUE = "#007bc3"
KORALL = "#ea5451"

font_family = "Charité Text Office"
plt.rcParams.update(
    {
        "font.family": "sans-serif",
        "font.sans-serif": [f"{font_family}"],
        "mathtext.fontset": "custom",
        "mathtext.rm": f"{font_family}",
        "mathtext.it": f"{font_family}:italic",
        "mathtext.bf": f"{font_family}:bold",
        "mathtext.bfit": f"{font_family}:bold:italic",
        "mathtext.sf": f"{font_family}",
    }
)

figure, axes = plt.subplots(ncols=1, nrows=1, figsize=(5, 5), dpi=300)
axes.scatter(
    data_frame["recall"][1:],
    data_frame["precision"][1:],
    label="Rater",
    color=SECOND_LBLUE,
)
axes.set_xlabel("Recall")
axes.set_ylabel("Precision")

axes.legend(
    loc="lower right",
)

axes.set_xlim(0, 1)
axes.set_ylim(0, 1)

axes.set_title(
    f"FMRI Open QC Project across {data_frame.height - 1} raters"
)

figure.savefig("precision-recall.svg", bbox_inches="tight", pad_inches=0, transparent=True)
```
